```python
# Construir lista de 200 usuarios
all_usuarios = []
all_categorias = []
# 200 = 2*23 + 7*22 → categorías 1 y 2 tienen 23, el resto 22
for cat in range(1, 10):
    n_cat = 23 if cat <= 2 else 22
    all_usuarios.extend(usuarios[cat][:n_cat])
    all_categorias.extend([cat] * n_cat)

all_usuarios = all_usuarios[:n]
all_categorias = all_categorias[:n]
```

```python
# Valores numéricos con 6 decimales
op        = np.round(np.random.uniform(30.0,  71.7, n), 6)
co        = np.round(np.random.uniform(7.9,   49.6, n), 6)
ex        = np.round(np.random.uniform(18.7,  59.8, n), 6)
ag        = np.round(np.random.uniform(9.3,   40.6, n), 6)
ne        = np.round(np.random.uniform(1.0,   24.0, n), 6)
wordcount = np.round(np.random.uniform(5.0,   217.2, n), 4)
```
